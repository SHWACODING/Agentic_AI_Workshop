### Imports ??

```python
from langgraph.graph import StateGraph, MessageGraph
from langchain_groq import ChatGroq
```

## Preparing LLM ??

```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ['GROQ_API_KEY'] = os.getenv("GROQ_API_KEY")
```

```python
llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    api_key=os.getenv("GROQ_API_KEY")
)
```

```python
response = llm.invoke("What is Qunatum Cryptography?")
```

```python
def AskQ1(input):
    llm = ChatGroq(model="llama-3.3-70b-versatile")
    response = llm.invoke(input)
    return response.content
```

```python
AskQ1("Explain Me Quantum LLM?")
```

```python
def Func2(input):
    upper_input = input.upper()
    return upper_input
```

```python
Func2("Explain me Quantum ML")
```

## Let's Build The Graph ??

```python
from typing import TypedDict

class AgentState(TypedDict):
    content: str
```

```python
workflow = StateGraph(str)

workflow.add_node("LLM_Agent", AskQ1)
workflow.add_node("Upper_str", Func2)
workflow.add_edge("LLM_Agent", "Upper_str")

workflow.set_entry_point("LLM_Agent")
workflow.set_finish_point("Upper_str")
```

```python
app = workflow.compile()
```

```python
from IPython.display import Image, display

try:
    display(Image(app.get_graph().draw_mermaid_png()))
except Exception as e:
    print(e)
```

```python
app.invoke("What are Topological Qubits?")
```
